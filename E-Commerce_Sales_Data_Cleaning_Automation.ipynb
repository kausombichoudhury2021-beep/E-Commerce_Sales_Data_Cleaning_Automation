```python
import pandas as pd
import numpy as np
import random

# Create 500 rows of messy data
data = {
    'Date': [random.choice(['2026-05-13', '13-05-2026', '2026/05/10', np.nan]) for _ in range(500)],
    'Product': [random.choice(['laptop', 'LAPTOP', 'Mouse', 'keyboard', None]) for _ in range(500)],
    'Sales': [random.choice(['300', '150', 'ERROR', '25', '1200', None]) for _ in range(500)],
    'Region': [random.choice(['North', 'South', 'East', 'West']) for _ in range(500)]
}

df_messy = pd.DataFrame(data)

# Add 100 duplicate rows to make it realistic
df_messy = pd.concat([df_messy, df_messy.iloc[:100]], ignore_index=True)

# Save it so you have the "Before" file
df_messy.to_csv('raw_sales_data_500.csv', index=False)

print("--- STEP 1 COMPLETE: Messy file 'raw_sales_data_500.csv' created! ---")
df_messy.head(10)
```

```python
import pandas as pd
import warnings

# Silences the red warning boxes
warnings.filterwarnings('ignore')

# Load the messy file we just made
df = pd.read_csv('raw_sales_data_500.csv')

# --- THE CLEANING PIPELINE ---
# 1. Fix Dates
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df['Date'] = df['Date'].ffill()

# 2. Fix Sales (Convert 'ERROR' to 0)
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce').fillna(0.0)

# 3. Fix Product names (Casing and missing values)
df['Product'] = df['Product'].str.title().fillna("General Merchandise")

# 4. Remove Duplicates
df = df.drop_duplicates()

# --- THE OUTPUT ---
# Save the final clean file
df.to_csv('final_cleaned_sales_data.csv', index=False)

print(f"--- STEP 2 COMPLETE: {len(df)} Unique Clean Rows Processed ---")
df.head(10)
```
